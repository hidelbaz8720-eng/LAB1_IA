```python
import string

text = "Hello World! I LOVE Python, and NLP is AMAZING... @2026"

text_lower = text.lower()
text_clean = text_lower.translate(str.maketrans("", "", string.punctuation))

print("Texte original :", text)
print("Texte nettoyé :", text_clean)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "I love learning artificial intelligence and programming with Python"

tokens = word_tokenize(text.lower())

stop_words = set(stopwords.words('english'))

print("Liste des stopwords :", sorted(stop_words))

filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]

# Ajouter un mot inutile à la liste des stopwords
stop_words.add("love")

filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]

print("Tokens originaux :", tokens)
print("Tokens après filtrage :", filtered_tokens)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["cars", "books", "playing", "studies"]

lemmas = [lemmatizer.lemmatize(word) for word in words]

print("Avant :", words)
print("Après :", lemmas)
```
